```python
import numpy as np
import matplotlib.pyplot as plt

months = np.arange(1, 13)

month_sin = np.sin(2 * np.pi * months / 12)
month_cos = np.cos(2 * np.pi * months / 12)

plt.figure()
plt.plot(months, month_sin, marker='o', label='sin encoding')
plt.plot(months, month_cos, marker='s', label='cos encoding')

plt.xlabel("Month")
plt.ylabel("Encoded Value")
plt.title("Cyclic Encoding of Month")
plt.xticks(months)
plt.legend()
plt.grid()

plt.savefig("cyclic_encoding_plot.png", dpi=300, bbox_inches='tight')
plt.show()
```
